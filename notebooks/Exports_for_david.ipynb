```python
import os
os.makedirs('./results/enrichment', exist_ok=True)

import pandas as pd

consensus_df = pd.read_csv('./results/DEG/consensus_DEGs.csv')

all_genes = consensus_df['Gene'].tolist()
up_genes = consensus_df[consensus_df['direction'] == 'Up']['Gene'].tolist()
down_genes = consensus_df[consensus_df['direction'] == 'Down']['Gene'].tolist()

print(f"All: {len(all_genes)}")
print(f"Up: {len(up_genes)}")
print(f"Down: {len(down_genes)}")

with open('./results/enrichment/all_consensus_genes.txt', 'w') as f:
    f.write('\n'.join(all_genes))

with open('./results/enrichment/up_consensus_genes.txt', 'w') as f:
    f.write('\n'.join(up_genes))

with open('./results/enrichment/down_consensus_genes.txt', 'w') as f:
    f.write('\n'.join(down_genes))

print("Files saved!")
```
