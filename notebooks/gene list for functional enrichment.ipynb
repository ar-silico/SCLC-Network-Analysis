```python
import pandas as pd

cluster1 = pd.read_csv('./results/PPI/cluster1_annotations.tsv', sep='\t')
cluster2 = pd.read_csv('./results/PPI/cluster2_annotations.tsv', sep='\t')
cluster3 = pd.read_csv('./results/PPI/cluster3_annotations.tsv', sep='\t')

# Save as text files for DAVID
with open('./results/enrichment/cluster1_genes.txt', 'w') as f:
    f.write('\n'.join(cluster1['#node'].tolist()))

with open('./results/enrichment/cluster2_genes.txt', 'w') as f:
    f.write('\n'.join(cluster2['#node'].tolist()))

with open('./results/enrichment/cluster3_genes.txt', 'w') as f:
    f.write('\n'.join(cluster3['#node'].tolist()))

print(f"Cluster 1: {len(cluster1)} genes")
print(f"Cluster 2: {len(cluster2)} genes")
print(f"Cluster 3: {len(cluster3)} genes")
print("All saved!")
```
