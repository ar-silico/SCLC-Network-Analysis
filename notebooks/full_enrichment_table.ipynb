```python
import pandas as pd
import numpy as np

categories_keep = ['GOTERM_BP_DIRECT', 'GOTERM_CC_DIRECT', 
                   'GOTERM_MF_DIRECT', 'KEGG_PATHWAY']

for name, filename in [('cluster1', 'cluster1_enrichment.csv'), 
                        ('cluster2', 'cluster2_enrichment.csv'), 
                        ('cluster3', 'cluster3_enrichment.csv')]:
    df = pd.read_csv(f'./results/enrichment/{filename}')
    
    filtered = df[df['Category'].isin(categories_keep)]
    
    filtered = filtered[filtered['Benjamini'] < 0.05]
    
    filtered = filtered.sort_values('Benjamini')
    
    filtered['-log10(Benjamini)'] = -np.log10(filtered['Benjamini'])
    
    filtered.to_csv(f'./results/enrichment/{name}_enrichment_filtered.csv', index=False)
    print(f"{name}: {len(filtered)} significant terms saved")
```
