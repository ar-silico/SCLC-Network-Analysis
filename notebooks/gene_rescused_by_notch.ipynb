```python
import pandas as pd

mcc_genes = ['BUB1B', 'KIF11', 'KIF2C', 'BUB1', 'ASPM', 
             'CDK1', 'TPX2', 'DLGAP5', 'CCNA2', 'CENPF']

betweenness_genes = ['MYC', 'IL6', 'CD44', 'BRCA1', 'PIK3R1', 
                     'CDK1', 'CCND1', 'TLR4', 'FLNA', 'KDR']

hub_genes_all = mcc_genes + [g for g in betweenness_genes if g not in mcc_genes]

notch_rescued = pd.read_csv('./results/DEG/notch_rescued_consensus_down.csv')
notch_suppressed = pd.read_csv('./results/DEG/notch_suppressed_consensus_up.csv')

print("Hub genes rescued by Notch:")
rescued_hubs = [g for g in hub_genes_all if g in notch_rescued['Gene'].values]
print(rescued_hubs)

print("\nHub genes suppressed by Notch:")
suppressed_hubs = [g for g in hub_genes_all if g in notch_suppressed['Gene'].values]
print(suppressed_hubs)
```
