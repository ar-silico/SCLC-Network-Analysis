```python
import os
os.makedirs('./results/PPI', exist_ok=True)

import pandas as pd

consensus_df = pd.read_csv('./results/DEG/consensus_DEGs.csv')
gene_list = consensus_df['Gene'].tolist()

with open('./results/PPI/consensus_genes_for_STRING.txt', 'w') as f:
    for gene in gene_list:
        f.write(gene + '\n')

print(f"Total genes: {len(gene_list)}")
print("File saved!")
```
