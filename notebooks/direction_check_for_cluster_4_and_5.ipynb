```python
import pandas as pd

cluster4 = pd.read_csv('./results/PPI/cluster4_annotations.tsv', sep='\t')
cluster5 = pd.read_csv('./results/PPI/cluster5_annotations.tsv', sep='\t')

consensus_df = pd.read_csv('./results/DEG/consensus_DEGs.csv')

for name, df in [('Cluster 4 (ECM-receptor)', cluster4), 
                 ('Cluster 5 (Complement/Coagulation)', cluster5)]:
    genes = df['#node'].tolist()
    directions = consensus_df[consensus_df['Gene'].isin(genes)]
    up = (directions['direction'] == 'Up').sum()
    down = (directions['direction'] == 'Down').sum()
    total = up + down
    print(f"{name}: {len(genes)} genes")
    print(f"  Upregulated: {up} ({up/total*100:.1f}%)")
    print(f"  Downregulated: {down} ({down/total*100:.1f}%)")
    print()
```

```python
rows = []
for name, df in [('Cluster 4 (ECM-receptor)', cluster4), 
                 ('Cluster 5 (Complement/Coagulation)', cluster5)]:
    genes = df['#node'].tolist()
    directions = consensus_df[consensus_df['Gene'].isin(genes)]
    up = (directions['direction'] == 'Up').sum()
    down = (directions['direction'] == 'Down').sum()
    total = up + down
    rows.append({
        'cluster': name,
        'total_genes': len(genes),
        'up': up,
        'down': down,
        'pct_up': round(up / total * 100, 1),
        'pct_down': round(down / total * 100, 1)
    })

summary_df = pd.DataFrame(rows)
summary_df.to_csv('./results/PPI/cluster4_5_direction_summary.csv', index=False)
summary_df
```
